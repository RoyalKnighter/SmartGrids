# Arbeitsauftrag: CIGRE - Mittelspannungsnetz

## Ziel
Erstellen Sie ein entsprechendes Modell des Mittelspannungs-Testnetzes CIGRE‑MV aus der pandapower Toolbox, führen Sie damit einen Lastfluss durch und stellen die wichtigsten Ergebnisse – Bus‑Spannungen und Leitungs‑Auslastungen – grafisch dar.  

## Der Auftrag gliedert sich in folgende Schritte:  

1. Netzaufbau – Erzeugen Sie das vorgegebene CIGRE‑MV‑Netz im pandapower‑Framework.  
2. Lastfluss – Lösen Sie das Netz mit dem Standard‑Lastfluss‑Algorithmus und prüfen Sie, dass das Ergebnis konvergiert.  
3. Ergebnis‑Extraktion – Ermitteln Sie aus den Ergebnis‑Tabellen die Spannungen an allen Sammelschienen (in p.u.) sowie die prozentuale Belastung jeder Leitung.  
4. Visualisierung – Fertigen Sie zwei übersichtliche Diagramme an, um die Bus‑Spannungen und die Leitungsbelastungen (mit einer Markierung bei 100 %) darzustellen.  
5. Dokumentation – Halten Sie schriftlich im Notebook die Vorgehensweise, die gewonnenen Kennzahlen und eine kurze Interpretation mit Literaturabgleich (z. B. ob Spannungen im zulässigen Bereich liegen und ob Leitungen überlastet sind) fest.
6. Ergänzen Sie das Netzwerk um Ihr in W2 erstelltes Ortsnetz mit Saisonaler Einspeisung und führen die Punkte 2-5 erneut aus.

## Ergebnis: Separates und vollständiges Jupyter‑Notebook, das den kompletten Workflow enthält und am Ende eine kurze textuelle Auswertung präsentiert.
